```python
#Create an Object of List
import sys
lst=[10,20,30,40,50,60]
print("Content of lst={} Type={}".format(lst,type(lst)))
print("Memory of Traditional Python List=",sys.getsizeof(lst))
```

```python
#Convert Traditional Python List into ndarray list
import numpy as np
lst=[10,20,30,40,50,60]
print("Content of lst=",lst)
a=np.array(lst)
print("Content of a={} Type={}".format(a,type(a)))
print("Memory of ndarray=",sys.getsizeof(a))
```

```python
#Create an Object of List---Memory Address
import sys
lst=[10,20,30,40,50,60,70,15,23,56,78,90,34,66,89,13,45,78]
for val in lst:
    print("Value={} Address of Value={} Address of lst={}".format(val,id(val),id(lst)))
```

```python
#Create an Object of list and Convert it into ndarray---Memory Address
import sys
lst=[10,20,30,40,50,60,70,15,23,56,78,90,34,66,89,13,45,78]
#Convert list into ndarray object
a=np.array(lst)
for val in a:
    print("Value={} Address of Value={} Address of a={}".format(val,id(val),id(a)))
```

```python
#Vector Based Operations
lst=[10,20,30,40]
print("Content of lst={} Type={}".format(lst,type(lst)))
#Increase Every Value of lst by 1
lst=lst+1
```

```python
#Vector Based Operations Not Possible By Default-
#Then How do u Get It on List---Vector Based Operations
lst=[10,20,30,40]
print("Content of lst={} Type={}".format(lst,type(lst)))
for i in range(len(lst)):
    lst[i]=lst[i]+1
print("Content of lst={} Type={}".format(lst,type(lst)))
```

```python
#Vector Based Operations Using ndarray Object
lst=[10,20,30,40]
print("Content of lst={} Type={}".format(lst,type(lst)))
#Convert the list Object into ndarray Object
a=np.array(lst)
print("Content of a={} Type={}".format(a,type(a)))
a=a+1   #Vector Based Operation
print("Content of a={} Type={}".format(a,type(a)))
```

```python
#Re-Shaping Mechanism in List
lst=[10,20,30,40]
print("Content of lst={} Type={}".format(lst,type(lst)))
lst.reshape(3,4)
```

```python
#To Reshape the Content of Traditional Python Iterable Object,
#We Must Convert Traditional Python Iterable Object into ndarray
lst=[10,20,30,40,50,60,70,80,90,15,25,35]
print("Content of lst={} Type={}".format(lst,type(lst)))
#Create list into ndarray
a=np.array(lst)
print("Content of a={} Type={}".format(a,type(a)))
```
