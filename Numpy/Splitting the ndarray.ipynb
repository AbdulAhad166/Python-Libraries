```python
#Splitting the 1d Array Data with Equal Splits---numpy.split()
import numpy as np
a=np.array([10,20,30,40,50,60,70,80])
print("Original Array Data")
print(a)
#Split the above 1D Array into 2 parts
parts=np.split(a,2)
print(parts,type(parts))
print("First Part of 1d Array={} Type={}".format(parts[0],type(parts[0])))
print("Second Part of 1d Array={} Type={}".format(parts[1],type(parts[1])))
```

```python
#OR Another Method for 1d Array 
import numpy as np
a=np.array([10,20,30,40,50,60,70,80])
print("Original Array Data")
print(a)
#Split the above 1D Array into 2 parts
parts=np.split(a,2)
for part in parts:
    print("Part={} Type={}".format(part,type(part)))
```

```python
#Syntax: varname=numpy.array_split(1DArray,Parts)
```

```python
#Splitting the 1d Array Data with Equal or unequal Splits---numpy.array_split()
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("Original Array Data")
print(a)
parts=np.array_split(a,2)
print(parts,type(parts))
print("First Part of 1d Array={} Type={}".format(parts[0],type(parts[0])))
print("Second Part of 1d Array={} Type={}".format(parts[1],type(parts[1])))
```

```python
#Or Another Method for 2D Array
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("Original Array Data")
print(a)
parts=np.array_split(a,4)
print(parts,type(parts))
for part in parts:
    print("Part={} Type={}".format(part,type(part)))
```

```python
#Splitting th 2D Array Data ------split()
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90,25,35,45]).reshape(4,3)
print("Original Array Data")
print(a)
parts=np.split(a,2,axis=0)
for part in range(len(parts)):
    print("Part:{}: \n{} Type={}".format(part+1,parts[part],type(parts[part])))
```
