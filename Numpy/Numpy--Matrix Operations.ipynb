```python
#dot Product of Matrices ---dot() 
import numpy as np
a=np.array([10,20,30,40]).reshape(2,2)
b=np.array([1,2,3,4]).reshape(2,2)
print("Matrix-A")
print(a)
print("Matrix-B")
print(b)
c=np.dot(a,b)
print("Matrix-C")
print(c)
```

```python
#dot Product of Matrices ---matmul()
import numpy as np
a=np.array([10,20,30,40]).reshape(2,2)
b=np.array([1,2,3,4]).reshape(2,2)
print("Matrix-A")
print(a)
print("Matrix-B")
print(b)
c=np.matmul(a,b)
print("Matrix-C")
print(c)
```
