```python
#Save the ndarray data into text File by using   numpy.savetxt()
import numpy as np
a=np.random.randint(2,10,size=(3,3))
print(a)
np.savetxt("F:\\Python-Libraries\\Numpy\\abc.data",a,delimiter=" ")
print("ndarray data saved in file---Verify")
```

```python
#Read the ndarray data from file into the main memory--numpy.loadtxt()
import numpy as np
a=np.loadtxt("F:\\Python-Libraries\\Numpy\\abc.data",delimiter=" ")
print(a,type(a),a.dtype)
b=a.astype(int)
print(b,type(b),b.dtype)
c=b.astype("i1")
print(c,type(c),c.dtype)
d=c.astype("i2")
print(d,type(d),d.dtype)
```

```python
#Save the ndarray data into .npy  File by using   numpy.save()
import numpy as np
a=np.random.randint(2,10,size=(3,3))
print(a)
np.save("F:\\Python-Libraries\\Numpy\\nit.npy",a)
print("ndarray Data Saved in npy File---Verify")
```

```python
#Save the ndarray data into .npy  File by using   numpy.save()
import numpy as np
a=np.random.randint(2,10,size=(3,3))
print(a)
b=np.load("F:\\Python-Libraries\\Numpy\\nit.npy")
print(b,type(b))
c=b.astype(float)
print(c,type(c),c.dtype)
d=c.astype("f2")
print(d,type(d),d.dtype)
e=d.astype("i1")
print(e,type(e),e.dtype)
```

```python
#Save the Mutiple ndarray Objects  data into  single file .npz  File by using   numpy.savez()
import numpy as np
a=np.random.randint(10,20,size=6)
b=np.random.randint(100,200,size=(3,3))
c=np.random.randint(30,50,size=(2,3,3))
np.savez("F:\\Python-Libraries\\Numpy\\hyd.npz",r=a,u=b,k=c)
print("Ndarray Objects Saved in npz File--Verify")
```

```python
#Read the Data from npz Files
import numpy as np
a=np.load("F:\\Python-Libraries\\Numpy\\hyd.npz")
print("Loaded Array")
print(a,type(a))
```

```python
print("First  Ndarray={} Type={}  Dim={}".format(a["r"],type(a["r"]), a["r"].ndim))
```

```python
print("Second  Ndarray\n{} Type={}  Dim={}".format(a["u"],type(a["u"]), a["u"].ndim))
```

```python
print("Third  Ndarray\n{} Type={}  Dim={}".format(a["k"],type(a["k"]), a["k"].ndim))
```

```python
#Save the Mutiple ndarray Objects  data into  single file .npz  File by using   numpy.savez()
import numpy as np
a=np.random.randint(10,20,size=6)
b=np.random.randint(100,200,size=(3,3))
c=np.random.randint(30,50,size=(2,3,3))
np.savez_compressed("F:\\Python-Libraries\\Numpy\\numpy.npz",r1=a,u2=b,k3=c)
print("Ndarray Objects Saved in npz File with Compressed--Verify")
```

```python
#Read the Compressed npz file data on the Basis of Key
a=np.load("F:\\Python-Libraries\\Numpy\\numpy.npz")
print("First  Ndarray={} Type={}  Dim={}".format(a["r1"],type(a["r1"]), a["r1"].ndim))
print("---------------------------------------")
print("Second  Ndarray\n{} Type={}  Dim={}".format(a["u2"],type(a["u2"]), a["u2"].ndim))
print("---------------------------------------")
print("Third  Ndarray\n{} Type={}  Dim={}".format(a["k3"],type(a["k3"]), a["k3"].ndim))
```
