```python
#Create an Object of 1darray---Discrete Differences
import numpy as np
a=np.array([10,28,35,49,5,6,70,85])
print("1D Array Data")
print(a)
b=np.diff(a)
print("Disecrete Difference of 1d Array=",b)
```

```python
#Create an Object of 1darray---Discrete Differences
import numpy as np
a=np.array([10,28,35,49,5,6,70,85])
print("1D Array Data")
print(a)
b=np.diff(a,n=1)
print("Discrete Difference of 1d Array=",b)
```

```python
#Create an Object of 1darray---Discrete Differences
import numpy as np
a=np.array([10,28,35,49,5,6,70,85])
print("1D Array Data")
print(a)
da1=np.diff(a,n=1)
print("Discrete Difference of 1d Array=",da1)
da2=np.diff(da1,n=1)
print("Discrete Difference of 1d Array=",da2)
```

```python
#Create an Object of 2darray---Discrete Differences
import numpy as np
a=np.array([10,28,35,49,5,6,70,85]).reshape(4,2)
print("2d Array Data")
print(a)
b=np.diff(a)
print("Discrete Difference of 2d Array\n",b)
```

```python
#Create an Object of 2darray---Discrete Differences-----Column-wise
import numpy as np
a=np.array([10,28,35,49,5,6,70,85]).reshape(4,2)
print("2d Array Data")
print(a)
b=np.diff(a,axis=1)
print("Discrete Difference of 2d Array\n",b)
```

```python
#Create an Object of 2darray---Discrete Differences-----Row-wise
import numpy as np
a=np.array([10,28,35,49,5,6,70,85]).reshape(4,2)
print("2d Array Data")
print(a)
b=np.diff(a,axis=0)
print("Discrete Difference of 2d Array\n",b)
```
