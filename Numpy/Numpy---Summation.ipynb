```python
#sum()
#Create an Object of 1d Array
import numpy as np
a=np.array([1,2,3])
b=np.sum(a)
print("Original Array Data")
print(a)
print("Sum=",b)
```

```python
#Create an Object of 1d Arrays
import numpy as np
a=np.array([1,2,3])
b=np.array([1,2,3])
c=np.sum([a,b])
print("Original 1d Array-1")
print(a)
print("Original 1d Array-2")
print(b)
print("Sum=",c)
```

```python
#Create an Object of 1d Arrays----Column-wise
import numpy as np
a=np.array([1,2,3])
b=np.array([1,2,3])
c=np.sum([a,b],axis=0)  #Column
print("original 1d Array-1")
print(a)
print("Original 1d Array-2")
print(b)
print("Sum=",c)
```

```python
#Create an Object of 1d Arrays----Row-wise
import numpy as np
a=np.array([1,2,3])
b=np.array([1,2,3])
c=np.sum([a,b],axis=1)  #Row
print("original 1d Array-1")
print(a)
print("Original 1d Array-2")
print(b)
print("Sum=",c)
```

```python
#Create an Object of 2d Array
import numpy as np
a=np.array([10,20,30,40,50,60]).reshape(2,3)
print("2D Array Data")
print(a)
b=np.sum(a)
print("Sum=",b)
```

```python
#Create an Object of 2d Array----Column-wise
import numpy as np
a=np.array([10,20,30,40,50,60]).reshape(2,3)
print("2D Array Data")
print(a)
b=np.sum(a,axis=0)
print("Sum=",b)
```

```python
#Create an Object of 2d Array----Row-wise
import numpy as np
a=np.array([10,20,30,40,50,60]).reshape(2,3)
print("2D Array Data")
print(a)
b=np.sum(a,axis=1)
print("Sum=",b)
```

```python
#Create an Object of 2d Arrays----Column-wise
import numpy as np
a=np.array([10,20,30,40,50,60]).reshape(2,3)
b=np.array([1,2,3,4,5,6]).reshape(2,3)
print("2D Array-1")
print(a)
print("2D Array-2")
print(b)
c=np.sum([a,b],axis=0)
print("Sum=",c)
```

```python
#cumsum()
#Create an Object of 1d Array
import numpy as np
a=np.array([1,2,3,4])
b=np.cumsum(a)
print("Original Array Data")
print(a)
print("Cummulative Sum=",b)
```

```python
#Create an Object of 1d Arrays---Column-wise
import numpy as np
a=np.array([1,2,3,4])
b=np.array([1,2,3,4])
c=np.cumsum([a,b],axis=0)
print("Original 1d Array-1")
print(a)
print("original 1d Array-2")
print(b)
print("Cummulative Sum")
print(c)
```

```python
#Create an Object of 1d Arrays----Row-wise
import numpy as np
a=np.array([1,2,3,4])
b=np.array([1,2,3,4])
c=np.cumsum([a,b],axis=1)
print("Original 1d Array-1")
print(a)
print("Original 1d Array-2")
print(b)
print("Cummulative Sum")
print(c)
```

```python
#Create an Object of 2d Array-----Column-wise
import numpy as np
a=np.array([10,20,30,40,50,60]).reshape(2,3)
print("2D Array Data")
print(a)
b=np.cumsum(a,axis=0)
print("Cummulative Sum")
print(b)
```

```python
#Create an Object of 2d Array-----Row-wise
import numpy as np
a=np.array([10,20,30,40,50,60]).reshape(2,3)
print("2d Array Data")
print(a)
b=np.cumsum(a,axis=1)
print("Cummulative Sum")
print(b)
```

```python
#Create an Object of 2d Arrays----Column-wise
import numpy as np
a=np.array([10,20,30,40,50,60]).reshape(2,3)
b=np.array([10,20,30,40,50,60]).reshape(2,3)
print("2d Array-1")
print(a)
print("2d Array-2")
print(b)
c=np.cumsum([a,b],axis=0)
print("Cummulative Sum")
print(c)
```
