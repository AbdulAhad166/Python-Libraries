```python
#Syntax---- varname=numpy.eye(N.M=None,k=0,dtype)
#Generates Unit OR Identity Matrix within Square OR Rectagle Matrix
```

```python
#Create an Object Using eye()
import numpy as np
a=np.eye(3)
print("Identity Matrix={} Type={}".format(a,type(a)))
print("Dimension of a={}".format(a.ndim))
```

```python
#Create an Object Using eye()
import numpy as np
a=np.eye(3,dtype=int)
print("Identity Matrix={} Type={}".format(a,type(a)))
print("Dimension of a={}".format(a.ndim))
```
