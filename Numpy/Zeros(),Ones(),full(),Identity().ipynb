```python
#Code for Building  a Matrix with dim of 2 with All entries as 0
import numpy as np
lst=[0,0,0,0]
a=np.array(lst)
print("Zero Matrix")
print(a)
print("Dimension of ndarray = {}".format(a.ndim))
```

```python
#Code for Building a Matrix with dim of 2 Dim With All entries as 0
import numpy as np
lst=[[0,0,0],[0,0,0],[0,0,0]]
a=np.array(lst)
print("Zero Matrix")
print(a)
print("Dimension of ndarray = {}".format(a.ndim))
```

```python
#Syntax---varname=numpy.zeros(shape,dtype)----zeros()
#Create an Object of ndarray with 6 zeros
import numpy as np
a=np.zeros(6)
print("Content of a={} Type={}".format(a,type(a)))
print("Dimension of ndarray = {}".format(a.ndim))
```

```python
#Create an Object of ndarray with 6 zeros---zeros()
import numpy as np
a=np.zeros(6,dtype=int)
print("Content of a={} Type={}".format(a,type(a)))
print("Dimension of ndarray={}".format(a.ndim))
```

```python
#Syntax---varname(shape,fill_value,dtype)----full()
#Create an Object of ndarray with 6 Entries as 3 dim ----full()
import numpy as np
a=np.full(6,3)
print("Content of a={} Type={}".format(a,type(a)))
print("Dimension of ndarray={}".format(a.ndim))
```

```python
#Create an Object of ndarray with 6 Entries as 3 dim -----full()
import numpy as np
a=np.full(6,3,dtype=float)
print("Content of a={} Type={}".format(a,type(a)))
print("Dimension of ndarray={}".format(a.ndim))
```

```python
#Create an Object of ndarray with 6 Entries as 3 dim -----full()
import numpy as np
a=np.full(shape=6,fill_value=3,dtype=float)
print("Content of a={} Type={}".format(a,type(a)))
print("Dimension of ndarray={}".format(a.ndim))
```

```python
#Create an Object of linspace() to Execute this Function
import numpy as np
a=np.linspace(0,10,num=10,retstep=True)
print("Content of a={} Type={}".format(a[0],type(a[0])))
print("Return Step of a={} Type={}".format(a[1],type(a[1])))
```

```python
#Create an Object Using linspace()
import numpy as np
a,rst=np.linspace(0,10,num=10,retstep=True)
print("Content of a={} Type={}".format(a,type(a)))
print("Return Step of rst={} Type={}".format(rst,type(rst)))
```

```python
#Create an Object Using linspace()
import numpy as np
a,rst=np.linspace(0,10,num=20,retstep=True)
print("Content of a={} Type={}".format(a,type(a)))
print("Return Step of rst={} Type={}".format(rst,type(rst)))
```
