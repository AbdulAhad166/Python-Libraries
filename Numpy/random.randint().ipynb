```python
#Creating for 2 Dimension array
import numpy.random as r
a=r.randint(1000,2000,(2,2))
print("Random Values={} Type={}".format(a,type(a)))
print("Dimension of a={} Shape={}".format(a.ndim,a.shape))
```

```python
#Creating for 2 Dimension array
import numpy.random as r
a=r.randint(1000,2000,(3,4))
print("Random Values={} Type={}".format(a,type(a)))
print("Dimension of a={} Shape={}".format(a.ndim,a.shape))
```

```python
print("Number of Rows={}".format(a.shape[0]))
print("Number of Columns={}".format(a.shape[1]))
```
