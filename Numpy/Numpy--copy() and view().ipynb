```python
#Create ndarray object of ndarray
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("ndarray Original Object={} Type={} id={}".format(a,type(a),id(a)))
```

```python
#Copy the Content of ndarray
b=np.copy(a)
print("ndarray Copied Object={} id={}".format(b,id(b)))
```

```python
#Do the Changes in Original Object and copied Object
a[0]=1000
b[-1]=2000
print("ndarray Original Object={} id={}".format(a,id(a)))
print("ndarray Copied Object={} id={}".format(b,id(b)))
```

```python
#Check the base for copied Object
print("Base for Copied Object={}".format(b.base))
```

```python
#Check the base for ndarray Object
print("Base for Original Object={}".format(a.base))
```

```python
#Alternativey we can write varname=ndarray Object.copy()
c=a.copy()
print("ndarray Original Obejct={} id={}".format(a,id(a)))
print("ndarray Copied Object={} id={}".format(c,id(c)))
```

```python
#Do the Changes in Original Object and Copied Object
a[0]=11
c[0]=22
print("ndarray Original Object={} id={}".format(a,id(a)))
print("ndarray Copied Object={} id={}".format(c,id(c)))
```

```python
#Implementation of view()
#Create ndarray object of ndarray
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("ndarray Object Original Data={} id={}".format(a,id(a)))
```

```python
# view the Orginal Ndarry Object 'a'
#b=np.view(a)----AttributeError: module 'numpy' has no attribute 'view'
```

```python
# view the Orginal Ndarry Object 'a'
b=a.view()    # here view() Present in ndarray object
print("ndarray View Object b={} id={}".format(b,id(b)))
```

```python
#Do the Changes in Original Object and View result with both Coiped and Viewd Object
a[0]=100
print("ndarray Original Object={} id={}".format(a,id(a)))
print("ndarray View Object={} id={}".format(b,id(b)))
```

```python
#Do the Changes in Viewed Object and View result with both Coiped and Viewd Object
b[-1]=34
print("ndarray Original Object={} id={}".format(a,id(a)))
print("ndarray Viewed Object={} id={}".format(b,id(b)))
```

```python
#Check the Base for Original Object
print("Base for Original Object={}".format(a.base))
```

```python
#Check the Base for Viewed Object
print("Base for Viewed Object={}".format(b.base))
```
