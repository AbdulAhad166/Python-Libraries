```python
#dot()---------------> Calculates Dot Product of Two Arrays
#Syntax: varname=numpy.dot(ndarray1,ndarray2)
```

```python
#Create an Object of 1d Arrays
import numpy as np
a=np.array([1,2,3])
b=np.array([4,5,6])
print("First 1d Array Data=",a)
print("Second 1d Array Data=",b)
c=np.dot(a,b)
print("Dot Product of 1D Arrays={}".format(c))
```

```python
#Create an Object of 2d Arrays
import numpy as np
a=np.array([1,3,5,2]).reshape(2,2)
b=np.array([2,4,6,3]).reshape(2,2)
print("First 2d Array\n",a)
print("Second 2d Array\n",b)
c=np.dot(a,b)
print("Dot Product of 2D Arrays\n",c)
```

```python
#Create an Object of 2d Arrays
import numpy as np
a=np.array([1,2,3,4,6,7]).reshape(3,2)
b=np.array([4,5,6,7]).reshape(2,2)
print("First 2d Array\n",a)
print("Second 2d Array\n",b)
c=np.dot(a,b)
print("Dot Product of 2D Arrays\n",c)
```

```python
# inner()------------> Calculates Inner Product of Two Arrays
#Syntax:  varname=numpy.inner(ndarray1,ndarray2) 
```

```python
#Create an Object of 1d Arrays---inner()
import numpy as np
a=np.array([1,3,5])
b=np.array([2,4,6])
print("First 2d Array=",a)
print("Second 2d Array=",b)
c=np.inner(a,b)
print("Inner Product of 1d Arrays=",c)
```

```python
#inner Product of 2D Arrays
import numpy as np
a=np.array([1,2,3,4]).reshape(2,2)
b=np.array([4,5,6,7]).reshape(2,2)
print("First 2d Array\n",a)
print("Second 2d Array\n",b)
c=np.inner(a,b)
print("Inner Product of 2d Arrays\n",c)
```

```python
#Note:  dot() of 1dArray is equal to inner() of 1dArray()
#But they are different in the 2-d.
```

```python
# outer()------------> Calculates Outer  Product of Two Arrays
#Syntax:  varname=numpy.outer(ndarray1,ndarray2)
```

```python
#Create an Object of 1d Arrays----outer()
import numpy as np
a=np.array([1,3,5])
b=np.array([2,4,6])
print("First 1d Array=",a)
print("Second 1d Array=",b)
c=np.outer(a,b)
print("Outer Product of 2d Arrays\n",c)
```

```python
#Create an Object of 1d Arrays----outer()
a=np.array([1,3,5,7]).reshape(2,2)
b=np.array([2,4,6,8]).reshape(2,2)
print("2-D Array a\n{}".format(a))
print("2-D Array b\n{}".format(b))
c=np.outer(a,b)
print("Outer Product of 2-D Arrays\n{}".format(c))
```

```python
#It is Present in "linalg" module 
# det()---------------> Calculates Determinant of Matrix (2dArray)
#Syntax:   varname=numpy.linalg.det(ndarray) here ndarray must be square matrix
#Formula for det= (a*d)-(b*c)
```

```python
#Creating an Object of 2D Array
import numpy as np
a=np.array([10,2,4,3]).reshape(2,2)
print("2d Array a\n",a)
b=np.linalg.det(a)
print("Det of 2D Matrix=",np.around(b,1))
```

```python
a=np.array([10,2,4,3,3,4]).reshape(2,3)
print("-"*50)
print("2-D Array a\n{}".format(a))
d=np.linalg.det(a) # LinAlgError: Last 2 dimensions of the array must be square
print("det of 2-D Matrix=",np.round(d,2))
```

```python
import numpy as np
a=np.array([10,2,4,3,3,4,1,6,4]).reshape(3,3)
print("-"*50)
print("2-D Array a\n{}".format(a))
d=np.linalg.det(a)
print("det of 2-D Matrix=",np.round(d,2))
```

```python
# solve()------------->Solves the Linear Equations in the form of Matrix format
#Syntax:  varname=numpy.linalg.solve(ndarray1,ndarray2)
```

```python
#Problem:   Solve the following Sysmtem Linear Equations
#          2x+4y=5
#          6x+8y=6    Find the values of x and y
```

```python
import numpy as np
#Create ndarray for x and y coff
a=np.array([[2,4],[6,8]])
b=np.array([5,6])
print("Matrix-A")
print(a)
print("Matrix-B")
print(b)
sol=np.linalg.solve(a,b)
print("Value of x=",np.around(sol[0],1))
print("Value of y=",np.around(sol[1],1))
```

```python
#Problem:   Solve the following Sysmtem Linear Equations
# 4x+2y=8
# 5x+3y=9    Find the values of x and y
```

```python
import numpy as np
a=np.array([4,2,5,3]).reshape(2,2)
b=np.array([8,9])
sol=np.linalg.solve(a,b)
print("Value of x=",np.around(sol[0],1))
print("Value of y=",np.around(sol[1],1))
```

```python
#Problem: Solve the following Sysmtem Linear Equations
# x+2y-z=4
# 3x-y+z=5
# 2x+3y+2z=7
```

```python
import numpy as np
a=np.array([[1,2,-1],[3,-1,1],[2,3,2]])
b=np.array([4,5,7])
sol=np.linalg.solve(a,b)
print("Values of x=",np.around(sol[0],2))
print("Values of y=",np.around(sol[1],2))
print("Values of z=",np.around(sol[2],2))
```

```python
# inv()----->Calculates of Inverse of a Martix (2D Array)
#Syntax: varname=numpy.linalg.inv(ndarrayobject)
```

```python
import numpy as np
a=np.array([4,2,5,3]).reshape(2,2)
print(a)
b=np.linalg.inv(a)
print("Inverse of Given Matrix")
print(b)
```

```python
import numpy as np
a=np.array([[1,2,-1],[3,-1,1],[2,3,2]])
print("Given Matrix")
print(a)
b=np.linalg.inv(a)
print("Inverse of Given Matrix")
for row in b:
    for val in row:
        print("{}".format(np.round(val,2)),end="  ")
    print()
```
