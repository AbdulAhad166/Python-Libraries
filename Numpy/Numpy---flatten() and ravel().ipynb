```python
#Create an Object of ndarray----2D Array
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("Matrix-A")
print(a)
print("Dimension of a={}".format(a.ndim))
print("Shape of a={}".format(a.shape))
#Flatten the ndarray Object --- flatten()
b=a.flatten()
print("Matrix-B")
print(b)
print("Dimension of b={}".format(b.ndim))
print("Shape of b={}".format(b.shape))
```

```python
#Create an Object of ndarray---3D
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90,15,25,35,45,75,45,55]).reshape(2,4,2)
print("Matrix-A")
print(a)
print("Dimension of a={}".format(a.ndim))
print("Shape of a={}".format(a.shape))
#Flatten the ndarray Object --- flatten()
b=a.flatten()
print("Matrix-B")
print(b)
print("Dimension of b={}".format(b.ndim))
print("Shape of b={}".format(b.shape))
```

```python
print("Address of a={}".format(id(a)))
print("Address of b={}".format(id(b)))
```

```python
#Check the Base of flatten Array
print("Flatten Array Base=",b.base)
```

```python
#Flatten the ndarray data by using ravel()
#Create an object of ndarray 
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("Matrix-A")
print(a)
print("Dimension of a={}".format(a.ndim))
print("Shape of a={}".format(a.shape))
#Flatten the ndarray Object --- ravel()
b=a.ravel()
print("Matrix-B")
print(b)
print("Dimension of b={}".format(b.ndim))
print("Shape of b={}".format(b.shape))
```

```python
print("Address of a={}".format(id(a)))
print("Address of b={}".format(id(b)))
```

```python
#Do the changes on Both ndarray Object
a[0][0]=11
print(a)
```

```python
print(a)
print("--------------------------")
print(b)
```

```python
#Do the changes on Both ndarray Object
b[-1]=99
print(b)
```

```python
print(a)
print("--------------------------")
print(b)
```

```python
#Update 0th Row of ndarray a
a[0:1,:]=[100,200,300]
print(a)
print("--------------------------")
print(b)
```

```python
b[-3:]=[700,800,900]
print(a)
print("--------------------------")
print(b)
```
