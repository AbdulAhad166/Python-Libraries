```python
#Numpy--Indexing Operation
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("Content of ndarray")
print(a)
print("Dimension of ndarray={} Shape={}".format(a.ndim,a.shape))
```

```python
#Indexing Operation  on 1D
```

```python
#Get 30 Element from 1dArray
print(a[2])
```

```python
#Get 90 Element from 1darray with +VE Index
print(a[len(a)-1])
```

```python
#Get 90 Element from 1darray with +VE Index
print(a[a.size-1])
```

```python
#Get The Element 10 From 1-d Array
print(a[0])
```

```python
print(a)
```

```python
#Get the Element 90 with -VE Index
print(a[-1])
```

```python
#Get the Element 10 with -VE Index
print(a[-a.size])
```

```python
#Get the Element 10 with -VE Index
print(a[-len(a)])
```

```python
#Indexing Operation  on 2D
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("Content of ndarray")
print(a)
print("Dimension of ndarray a={}".format(a.ndim))
print("Shape of ndarray a={}".format(a.shape))
```

```python
#Indexing Operation  on 3D
import numpy as np
a=np.arange(3*3*3).reshape(3,3,3)
print("Content of ndarray\n{}".format(a))
print("Dimension of ndarray={}".format(a.ndim))
print("Shape of ndarray={}".format(a.shape))
```

```python
#Get First Matrix From 3D
print(a[0])
```

```python
#Get Second Matrix From 3D
print(a[1])
```
