```python
#Create an object of ndarray Array
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8])
print("1d Array Data")
print(a)
```

```python
#Find Max Element from 1D Array
#Find Highest Element in 1DArray---amax()
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8])
print("1d Array Data")
print(a)
maxv=np.amax(a)
print("Maximum Element=",maxv)
```

```python
#Find Max Element from 2D Array
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8]).reshape(3,3)
print("2d Array Elements")
print(a)
maxv=np.amax(a)
print("Max Element=",maxv)
```

```python
#Find Max Element from 2D Array in Column-wise
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8]).reshape(3,3)
print("2d Array Elements")
print(a)
maxv=np.amax(a)
colmaxv=np.amax(a,axis=0)  # Here axis=0 Represents Columns
print("Max Element=",maxv)
print("Column Max Elements=",colmaxv)
```

```python
#Find Max Elements from 2D Array in Row-wise
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8]).reshape(3,3)
print("2d Array Elements")
print(a)
maxv=np.amax(a)
rowmaxv=np.amax(a,axis=1)  # Here axis=1 Represents Rows
print("Max Element=",maxv)
print("Row Max Elements=",rowmaxv)
```

```python
#Syntax:  varname=numpy.amin(ndarray,axis)
```

```python
#Find Min Element from 1D Array
#Find Smallest Element in 1DArray---amin()
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8])
print("1d Array Elements")
print(a)
minv=np.amin(a)
print("Min Element=",minv)
```

```python
#Find Min Element from 1D Array in Column-wise
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8]).reshape(3,3)
print("2d Array Elements")
print(a)
minv=np.amin(a)
colminv=np.amin(a,axis=0) # Here axis=0 Represents Cols
print("Min Element=",minv)
print("Column Min Values=",colminv)
```

```python
#Find Min Element from 1D Array in Row-wise
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8]).reshape(3,3)
print("2d Array Elements")
print(a)
minv=np.amin(a)
rowminv=np.amin(a,axis=1) # Here axis=1 Represents Rows
print("Min Element=",minv)
print("Row Min Values=",rowminv)
```

```python
#Find Min Element from 3D Array in Row-wise
import numpy as np
a=np.array([10,2,13,4,14,5,15,18,8,12,9,10,25,8,4,14]).reshape(2,4,2)
print("3d Array Elements")
print(a)
maxv=np.amax(a)
print("Max Element=",maxv)
Fcolmaxv=np.amax(a[0],axis=0)
Scolmaxv=np.amax(a[1],axis=0)
print("First Matrix Col Max Vals\n",Fcolmaxv)
print("Second Matrix Col Max Vals\n",Scolmaxv)
```

```python
#Finding the Mean of ndarray Elements--mean()
#Syntax: varname=numpy.mean(ndarray,axis)
```

```python
#Consider ndarray and find mean
import numpy as np
a=np.array([2,3,1,4])
print("1D Array Data")
print(a)
mv=np.mean(a)
print("Mean of 1DArray=",mv)
```

```python
#Consider 2darray and find mean with row and column
import numpy as np
a=np.array([2,3,1,4]).reshape(2,2)
print("2D Array Data")
print(a)
mv=np.mean(a)
colmean=np.mean(a,axis=0) # Here axis=0 Represents Cols
rowmean=np.mean(a,axis=1) # Here axis=1 Represents Rows
print("Mean of 2D Array=",mv)
print("Column-wise mean=",colmean)
print("Row-wise mean=",rowmean)
```

```python
#Consider 2darray and find mean with row and column
import numpy as np
a=np.random.randint(100,200,size=(4,7))
print("2D Array Data")
print(a)
mv=np.mean(a)
colmean=np.mean(a,axis=0)
rowmean=np.mean(a,axis=1)
print("Mean of 2D Array=",mv)
print("Column-wise mean=",colmean)
print("Row-wise mean=",rowmean)
```

```python
#Consider the following 2darray and find mean
import numpy as np
a=np.random.randint(2,10,size=(3,3))
print("2D Array Data")
print(a)
```

```python
#Find the Mean of Opp Principle Diagnoal Elements
import numpy as np
pd=np.diag(a)
print("Principle Diagonal Elements=",pd)
pdmv=np.mean(pd)
print("Mean of Principle Diagonal Elements=",pdmv)
```

```python
#Find the Mean of Opposite Principle Diagonal Elements
import numpy as np
a=np.random.randint(2,10,size=(3,3))
print("2D Array Data")
print(a)
b=a[[0,1,2],[2,1,0]]
print("Opposite Principle Diagonals")
mv=np.mean(b)
print("Mean of Opposite Principle Diagonal Elements=",mv)
```

```python
#Find  the Median of ndarray---median()
a=np.random.randint(1,50,size=7)
print("1D Array Data")
print(a)
```

```python
#Median  of 1D Array of odd Number of Elements
mdv=np.median(a)
print("Median of 1D Array=",mdv)
```

```python
#Median  of 1D Array of Even Number of Elements
a=np.random.randint(12,50,size=8)
print("1D Array Data")
print(a)
mdv=np.median(a)
print("Median of 1D Array=",mdv)
```

```python
a=np.random.randint(12,50,size=(3,3))
print("2D Array Data")
print(a)
```

```python
#Median of 2D
import numpy as np
mv=np.median(a)
colmv=np.median(a,axis=0)
rowmv=np.median(a,axis=1)
print("Median Value=",mv)
print("Column Median=",colmv)
print("Row Median=",rowmv)
```

```python
#Variance of ndarray---1D Array
import numpy as np
a=np.array([2,3,1,4])
print("1D Array Data")
print(a)
v=np.var(a)
print("Variance of a=",v)
```

```python
#Variance of ndarray----2D Array
import numpy as np
a=np.array([2,3,1,4]).reshape(2,2)
print("2D Array Data")
print(a)
v=np.var(a)
colv=np.var(a,axis=0)
rowv=np.var(a,axis=1)
print("Variance of a=",v)
print("Column Variance=",colv)
print("Row Variance=",rowv)
```

```python
#Standard Deviation= sqrt(var) of ndarray----1D Array
import numpy as np
a=np.array([2,3,1,4])
print("1D Array Data")
print(a)
d=np.std(a)
print("Standard Deviation=",d)
```

```python
#Standard Deviation= sqrt(var) of ndarray----2D Array
import numpy as np
a=np.array([2,3,1,4]).reshape(2,2)
print("1D Array Data")
print(a)
d=np.std(a)
cold=np.std(a,axis=0)
rowd=np.std(a,axis=1)
print("Standard Deviation of a=",d)
print("Column Standard Deviation=",cold)
print("Row Standard Deviation=",rowd)
```
