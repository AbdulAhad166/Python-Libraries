```python
#Create an Object of 1D Array
import numpy as np
a=np.array([1,2,3,4])
b=np.prod(a)
print("Original Array Data")
print(a)
print("Product=",b)
```

```python
#Create an Object of 1D Arrays
import numpy as np
a=np.array([1,2,3,4])
b=np.array([1,2,3,4])
c=np.prod([a,b])
print("Original Array Data-1")
print(a)
print("Original Array Data-2")
print(b)
print("Product=",c)
```

```python
#Create an Object of 1D Arrays---Column-wise
import numpy as np
a=np.array([1,2,3])
b=np.array([1,2,3])
c=np.prod([a,b],axis=0)
print("Original Array Data-1")
print(a)
print("Original Array Data-2")
print(b)
print("Product=",c)
```

```python
#Create an Object of 1D Arrays----Row-wise
import numpy as np
a=np.array([1,2,3])
b=np.array([1,2,3])
c=np.prod([a,b],axis=1)
print("Original Array Data-1")
print(a)
print("Original Array Data-2")
print(b)
print("Product=",c)
```

```python
#Create an Object of 2D Array
import numpy as np
a=np.array([1,2,3,4,5,6]).reshape(2,3)
print("Original Array Data")
print(a)
b=np.prod(a)
print("Product=",b)
```

```python
#Create an Object of 2D Array----Column-wise
import numpy as np
a=np.array([1,2,3,4,5,6]).reshape(2,3)
print("Original Array Data")
print(a)
b=np.prod(a,axis=0)
print("Product=",b)
```

```python
#Create an Object of 2D Array
import numpy as np
a=np.array([1,2,3,4,5,6]).reshape(2,3)
print("Original Array Data")
print(a)
b=np.prod(a,axis=1)
print("Product=",b)
```

```python
#Create an Object of 2D Arrays----Column-wise
import numpy as np
a=np.array([1,2,3,4]).reshape(2,2)
b=np.array([1,2,3,4]).reshape(2,2)
c=np.prod([a,b],axis=0)
print("Original Array Data-1")
print(a)
print("Original Array Data-2")
print(b)
print("Product\n",c)
```

```python
#Create an Object of 2D Arrays----Row-wise
import numpy as np
a=np.array([1,2,3,4]).reshape(2,2)
b=np.array([1,2,3,4]).reshape(2,2)
c=np.prod([a,b],axis=1)
print("Original Array Data-1")
print(a)
print("original Array Data-2")
print(b)
print("Product\n",c)
```

```python
#cumprod()
#Create an object of 1darray
import numpy as np
a=np.array([1,2,3,4])
b=np.cumprod(a)
print("Original 1d Array")
print(a)
print("Cumulative Product=",b)
```
