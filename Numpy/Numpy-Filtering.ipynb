```python
#Create 1D Array with Random 12 Integer Values
from numpy import random as r
a=r.randint(25,55,size=12)
print("Random Elements={}".format(a))
```

```python
#Get the Factors of 5 --Normal Python Code
lst=[]  #Empty List
for val in a:
    if (val%5==0):
        lst.append(int(val))
print("Factors of 5=",lst)
```

```python
#Get the Factors of 5 --Normal Python Code---List Comprehension
lst=[int(val) for val in a if(val%5==0)]
print("Factors of 5=",lst)
```

```python
#Create 1D Array with Random 12 Integer Values
from numpy import random as r
a=r.randint(25,55,size=12)
print("Random Elements={}".format(a))
```

```python
#Get all Factors of 5
#Step-1: Prepare the Condition with ndarray object
ba=a%5==0
print("Boolean Array={} Type={}".format(ba,type(ba)))
#Step-2: Pass the Boolean Array to the ndarray Object
print("Factors of 5")
print(a[ba])
```

```python
#Create 1D Array with Random 12 Integer Values
from numpy import random as r
a=r.randint(25,55,size=12)
print("Random Elements")
print(a)
#Get All Factors of 3
#Step-1: Prepare the Condition with ndarray Object
ba=a%3==0
print("Boolean Array={} Type={}".format(ba,type(ba)))
#Step-2: Pass the Boolean Array to the ndarray Object
print("Factors of 3")
print(a[ba])
```

```python
#Create 1D Array with Random 12 Integer Values
from numpy import random as r
a=r.randint(25,55,size=12)
print("Random Elements")
print(a)
#Get All Factors of 5 & 3
#Step-1: Prepare the condition with ndarray Object
ba=(a%3==0) & (a%5==0)
print("Boolean Array={} Type={}".format(ba,type(ba)))
#Step-2: Pass the Boolean Array to the ndarray Object
print("Factors of 5 & 3")
print(a[ba])
```

```python
#Create 1D Array with Random 15 Integer Values
from numpy import random as r
a=r.randint(25,55,size=15)
print("Random Elements")
print(a)
#Get all Factors of 5 and 3
print("------------------------------------------------")
print("Factors of 5 and 3")
print(a[(a%3==0)&(a%5==0)])
```
