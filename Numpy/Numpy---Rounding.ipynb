```python
#trunc()
#Create an Object of 1darray
import numpy as np
a=np.array([-3.1666, 3.6667,2.999,1.999,3.999,5.6666])
print("Original Array Data=",a)
b=np.trunc(a)
print("Truncated ndarray Object=",b)
```

```python
#Create an Object of 2darray
import numpy as np
a=np.array([-3.1666, 3.6667,2.999,1.999,3.999,5.6666]).reshape(3,2)
print("Original Array Data\n",a)
b=np.trunc(a)
print("Truncated ndarray Object\n",b)
```

```python
#fix()
#Create an Object of 1darray
import numpy as np
a=np.array([-3.1666, 3.6667,2.999,1.999,3.999,5.6666]).reshape(3,2)
print("Original Array Data\n",a)
b=np.fix(a)
print("Truncated ndarray Object\n",b)
```

```python
#Rounding the Decimals By Using around()
import numpy as np
arr=np.around(3.1666,2)
print(arr)
```

```python
#Create an Object of 1darray
import numpy as np
a=np.array([-3.1666, 3.6667,2.995,1.999,3.919,5.6646])
print("Original Array Data\n",a)
b=np.around(a,2)
print("Rounded Array Values\n",b)
```

```python
#Create an Object of 2d Array
import numpy as np
a=np.array([-3.1666, 3.6667,2.995,1.999,3.919,5.6646]).reshape(3,2)
print("Original Array Data\n",a)
b=np.around(a,2)
print("Rounded Array Values\n",b)
```

```python
#Floor the Elements of Array---floor()
import numpy as np
a=np.floor([-3.1666, 3.6667])
print(a)
```

```python
#Create an Object of 1darray----floor()
import numpy as np
a=np.array([-3.1666, 3.6667,2.995,1.999,3.919,-5.6646])
print("Original Array Data\n",a)
b=np.floor(a)
print("Floor Array Values\n",b)
```

```python
#The ceil() function rounds-off decimal to nearest Highest OR  upper integer.
```

```python
#Create an Object of 1darray
import numpy as np
a=np.ceil([-3.1666, 3.6667])
print(a)
```

```python
#Create an Object of 1darray
import numpy as np
a=np.array([-3.1666, 3.6667,2.995,1.999,3.00001,-5.6646])
print("Original Array Data\n",a)
b=np.ceil(a)
print("Ceil Array Values\n",b)
```

```python
#Create an Object of 2darray
import numpy as np
a=np.array([-3.1666, 3.6667,2.995,1.999,3.00001,-5.6646]).reshape(3,2)
print("Original Array Data\n",a)
b=np.ceil(a)
print("Ceil Array Values\n",b)
```
