```python
#Using arange()----varname=numpy.arange(Value)
import numpy as np
a=np.arange(10)
print("Content of a={} Type={}".format(a,type(a)))
print("Dimension of ndarray={}".format(a.ndim))
print("Shape of ndarray={}".format(a.shape))
print("Size of ndarray={}".format(a.size))
```

```python
#Using arange()----varname=numpy.arange(Value)----2D
import numpy as np
a=np.arange(10).reshape(5,2)
print("content of a\n{}  Type={}".format(a,type(a)))
print("Dimension ndarray ={}".format(a.ndim))  
print("Shape of ndarray={}".format(a.shape))   
print("Size of ndarray={}".format(a.size)) 
```
