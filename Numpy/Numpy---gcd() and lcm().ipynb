```python
#GCD of ndarray Values
import numpy as np
n1=10
n2=3
res=np.gcd(n1,n2)
print("GCD({},{})={}".format(n1,n2,res))
```

```python
#GCD of ndarray Values
import numpy as np
a=np.array([20,4,3,10,15,19])
res=np.gcd.reduce(a)
print("GCD({})={}".format(a,res))
```

```python
#LCM of ndarray Values
import numpy as np
n1=10
n2=3
res=np.lcm(n1,n2)
print("GCD({},{})={}".format(n1,n2,res))
```
