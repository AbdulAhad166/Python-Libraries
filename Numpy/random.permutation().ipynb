```python
#Re-Arrange The Elements of ndarray By Using permutation()
from numpy import random as r
a=r.randint(10,21,size=6)
print("Original Content={} Type={} id={}".format(a,type(a),id(a)))
ra=r.permutation(a)
print("Original Content of ndarray a after permutation()={} id={}".format(a,id(a)))
print("Re-Arranged Elements of ndarray={} id={}".format(ra,id(ra)))
```

```python
#Re-Arrange the Elements of ndarray By Using permutation()
from numpy import random as r
a=r.randint(10,21,size=6)
print("Original Content={} Type={} id={}".format(a,type(a),id(a)))
for i in range(1,6):
    ra=r.permutation(a)
    print("{} Time: Re-Arranged Elements={} id={}".format(i,ra,id(ra)))
```

```python
#Re-Arrange the Elements of ndarray By Using permutation()
import numpy as np
a=np.array(list("MISSISSIPPI"))
print("Original Content={} Type={} id={}".format(a,type(a),id(a)))
for i in range(1,len(a)+1):
    ra=np.random.permutation(a)
    print("{} Time: Re-Arranged Elements={} id={}".format(i,"".join(ra),id(ra)))
```
