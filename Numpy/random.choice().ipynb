```python
#Generate 5 Random Integer Values
from numpy import random as r
a=r.randint(10,30,size=9)
print("Random Values={} Type={}".format(a,type(a)))
#Select One Value from List of Random Samples
cv=r.choice(a)
print("Random Choiced Values={}".format(cv))
```

```python
#Getting Random Alphabets Using choice()
from numpy import random as r
upalphs=[chr(val) for val in range(65,91)]
lowalphs=[chr(val) for val in range(97,123)]
allalphs=upalphs+lowalphs
#Select 3 Alphabets from List Object as Random Samples
cv=r.choice(allalphs,3)
print("Random Choiced Value={} Type={}".format(cv,type(cv)))
```

```python
#Getting Random Alphabets Using choice()
from numpy import random as r
upalphs=[chr(val) for val in range(65,91)]
lowalphs=[chr(val) for val in range(97,123)]
allalphs=upalphs+lowalphs
#Select Alphabets from list Object as Random Samples
cv=r.choice(allalphs,6)
print("Random Choiced Value={} Type={}".format("".join(cv),type(cv)))
```

```python
import numpy as np
a=np.arange(10)
print("Original Content={} id={}".format(a,id(a)))
#Re-Arrange the Values in Same Place (in-place)
np.random.shuffle(a)
print("Re-Arranged Values={} id={}".format(a,id(a)))
```

```python
import numpy as np
a=np.arange(10)
print("Original Content={} id={}".format(a,type(a)))
#Re-Arrange the Values in Same Place (in-place)
for i in range(1,6):
    np.random.shuffle(a)
    print("{} Time: Re-Arranged Content of a={} id={}".format(i,a,id(a)))
```

```python
import numpy as np
a=np.array((list("MISSISSIPPI")))
print("Original Content of a={} id={} Type={}".format(a,id(a),type(a)))
for i in range(1,len(a)+1):
    np.random.shuffle(a)
    print("{} Time: Re-Arranged Content of a={} id={}".format(i,"".join(a),id(a)))
```
