```python
#Numpy--Slicing Operations on 1-D
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("Content of ndarray a={}".format(a))
print("Dimension of ndarray={}".format(a.ndim))
print("Shape of ndarray={}".format(a.shape))
```

```python
#Get 10 20 30 40  from 1D Array
b=a[0:4]
print(b,type(b))
```

```python
print(a)
```

```python
#Get the Elements 60 70 80 90
print(a[5:9])
```

```python
#Get the Elements 60 70 80 90-----Another Way
print(a[5:])
```

```python
#Get the Elements 30,40,50,60 with -VE Slicing
print(a[-7:-3])
```

```python
print(a)
```

```python
#Get 10 30 50 70 70 
print(a[::2])
```

```python
#Get 20 40 60 80 
print(a[1::2])
```

```python
print(a)
```

```python
#Get 10 40 70 
print(a[::3])
```

```python
#Get 90,80,70,60 in Backward Direction 
print(a[-1:-5:-1])
```

```python
#Get 90 70 50 30  10 in Backward Direction with -VE Step
print(a[::-2])
```

```python
#Get 90 70 50 30  10 in Backward Direction with +VE Step
print(a[::2][::-1])
```

```python
#Numpy-Slicing Operations on 2-d
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90,34,1,11,13,14,15,16]).reshape(4,4)
print("Content of ndarray\n{}".format(a))
print("Dimension of ndarray={}".format(a.ndim))
print("Shape of ndarray={}".format(a.shape))
```

```python
#Get [60,70,80] from 2D Array
a[1:2,1:4]
```

```python
#Get [60,70,80] from 2D Array---Another Way
a[1:2,1:]
```

```python
#Get [60,70,80] from 2D Array with -ve Indices
a[-3:-2,-3:]
```

```python
#Get 60,70,34,1 from 2D Array
a[1:3,1:3]
```

```python
#Get 60,70,34,1 from 2D Array---Another Way
a[1:3:1,1:3:1]
```

```python
#Get 60,70,34,1 from 2D Array---Another Way
a[1:3:,1:3:]
```

```python
print(a)
```

```python
print(a[1:,1:4])
```

```python
print(a[1:,1:3])
```

```python
a[1::,1:3:1]
```

```python
a[1::1,1:3:1]
```

```python
print(a)
```

```python
#Get 50,60,70,80,13,14,15,16
a[1::2,:]
```

```python
#Create 3D Array with Random Integer Values with shape(3,3,3)
import numpy.random as r
a=r.randint(10,20,(3,3,3))
print("3-d Array")
print(a)
print("Shape of a={}".format(a.shape))
print("Dimension of a={}".format(a.ndim))
```

```python
#Get First Two Matrices from 3D
print(a[0:2])
```

```python
#Get First and Last Matrix
print(a[0::2])
```

```python
print(a)
```

```python
#Get All 0th row of all Columns of all Matrices
print(a[:,0:1,0:3])
```

```python
#Get All 0th Col of all Rows of all Matrices
print(a[:,:,0:1])
```

```python
print(a)
```

```python
#Get First and Last Matrices 1st Row Elements of Corresposnding Matrices
print(a[::2,1:2,::])
```

```python
print(a[::2,2::,1::])
```

```python
print(a[::2,:2:2,2::2])
```

```python
print(a[2::,2::,::2])
```
