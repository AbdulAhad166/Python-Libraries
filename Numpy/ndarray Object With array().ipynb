```python
#Creating an Object of ndarray Using array()---dtype
import numpy as np
lst=[10,20,30,40,50,60]
print("Content of lst={} Type={}".format(lst,type(lst)))
a=np.array(lst,float).reshape(3,2)
print("Content of ndarray={} Type={}".format(a,type(a)))
print("Dimension of ndarray={}".format(a.ndim))
print("Shape of ndarray={}".format(a.shape))
print("Size of ndarray={}".format(a.size))
print("Data Type Elements of ndarray={}".format(a.dtype))
```

```python
#Creating an Objct of ndarray Using array()----Nested List
import numpy as np
lst=[[[10,20],[30,40]],[[50,60],[70,80]]]
print("Content of list={}  Type list={}".format(lst,type(lst)))
a=np.array(lst,float)
print("Content of ndarray\n{}  Type={}".format(a,type(a)))
print("Dimension ndarray ={}".format(a.ndim))
print("Shape of ndarray={}".format(a.shape)) 
print("Size of ndarray={}".format(a.size)) 
print("Data type Elements of ndarray={}".format(a.dtype)) 
```
