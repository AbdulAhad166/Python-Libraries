```python
#Insert a Row in 2DArray
#Create an object of ndarray 
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("2ddarray Data")
print(a)
```

```python
#Insert row [11,22,33] at 1 Row of 2d
a=np.insert(a,1,[11,22,33],axis=0)
```

```python
print(a)
```

```python
#Insert at 1 column [55,65,75,85] of 2d
a=np.insert(a,1,[55,65,75,85],axis=1)
```

```python
print(a)
```

```python
#Insert row [10,20,30,40] at Last in 2d
a=np.insert(a,len(a),[10,20,30,40],axis=0)
```

```python
print(a)
```

```python
#Insert Column [11,21,31,41,51] at Last in Begining
a=np.insert(a,0,[11,21,31,41,51],axis=1)
```

```python
#To delete the Element from 1D Array-we use delete()
#Syntax-1:    numpy.delete(1dArray, Index)
#Syntax-2:    numpy.delete(1dArray, [Index1,Index2,...,Index-n] )
```

```python
#Create an Object of 1D Array
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("Original Array Data")
print(a)
#Delete the value 30 in the ndarray
a=np.delete(a,2)
print("1D Array Data After Delete")
print(a)
```

```python
#Create an Object of 1D Array 
import numpy as np
a=np.array([10,20,30,50,60,70,80,90])
print("Original 1D Array Data")
print(a) # [10 20 30 50 60 70 80 90]
#Delete  30 , 70 and 80 from 1D
a=np.delete(a,[2,-3,-2])
print("1D Array Data after delete")
print(a)
```

```python
#Create an Object of 2D Array
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("Original 2D Array Data")
print(a)
#Delete 1 Row from 2D
a=np.delete(a,1,axis=0)
print("2D Array Data After Row Deletion")
print(a)
```

```python
#Create an Object of 2D Array
import numpy as np
a=np.arange(5*5).reshape(5,5)
print("Original 2D Array Data")
print(a)
#Delete 1 Row, 2row and last row  from 2D 
a=np.delete(a,[1,2,-1],axis=0)
print("2D Array Data After Row Deletion")
print(a)
```

```python
#Create an Object of 2D Array
import numpy as np
a=np.arange(5*5).reshape(5,5)
print("Original 2D Array Data")
print(a)
#Delete 1 Column from 2D
a=np.delete(a,1,axis=1)
print("2D Array Data After Column Deletion")
print(a)
```

```python
#Create an Object of 2D Array
import numpy as np
a=np.arange(5*5).reshape(5,5)
print("Original Array Data")
print(a)
#Delete 0th,last Two Columns from 2D 
a=np.delete(a,[0,-2,-1],axis=1)
print("2D Array Data After Columns Deletion")
print(a)
```
