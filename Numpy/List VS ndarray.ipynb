```python
#Complex Math Operations With List
lst1=[10,20,30,40]
lst2=[15,34,1,9]
print("List-1=",lst1)
print("List-2=",lst2)
lst3=lst1+lst2
print("List-3=",lst3)
```

```python
lst3=[]
for x,y in zip(lst1,lst2):
    lst3.append(x+y)
print("List-1=",lst1)
print("List-2=",lst2)
print("List-3=",lst3)
```

```python
#Complex Math Operations With ndarray
import numpy as np
lst1=[10,20,30,40]
lst2=[15,34,1,9]
a=np.array(lst1)
b=np.array(lst2)
print("a=",a)
print("b=",b)
c=a+b
print("c=",c)
```
