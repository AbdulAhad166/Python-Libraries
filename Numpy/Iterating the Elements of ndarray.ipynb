```python
#By Using Looping Statements To Display the Data of ndarray
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("1-d Array Data={} Dimension={}".format(a,a.ndim))
for val in a:
    print("\t {}".format(val))
```

```python
#By Using Looping Statements To Display the Data of ndarray for 2-d
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("2-d Array Data\n{} Dimension={}".format(a,a.ndim))
for row in a:
    for val in row:
        print("\t{}".format(val))
```

```python
#By Using Looping Statements To Display the Data of ndarray for 3-d
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90,34,1,55]).reshape(2,3,2)
print("3-d Array Data\n{} Dimension={}".format(a,a.ndim))
for matrix in a:
    for row in matrix:
        for val in row:
            print("\t{}".format(val))
        print("-------------------------")
```

```python
#By Using Looping Statements To Display the Data of ndarray for 4-d
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90,34,1,55,16,18,19,11]).reshape(2,2,2,2)
print("4-d Array Data\n{} Dimension={}".format(a,a.ndim))
for cluster in a:
    for matrix in cluster:
        for row in matrix:
            for val in row:
                print("\t {}".format(val))
```

```python
#By Using numpy.ndenumerate() To Display the Data of ndarray for 1-d
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("1-d Array Data={} Dimension={}".format(a,a.ndim))
for val in np.ndenumerate(a):
    print("\t {}--->{}".format(val[0],val[1]))
```

```python
#By Using numpy.ndenumerate() To Display the Data of ndarray for 3-d
import numpy as np
a=np.array([10,20,30,40,50,60,70,80]).reshape(2,2,2)
print("3-d Array Data\n{} Dimension={}".format(a,a.ndim))
for val in np.ndenumerate(a):
    print("\t Matrix:{} Row:{} Column:{} Value:{}".format(val[0][0],val[0][1],val[0][2],val[1]))
```

```python
#By Using numpy.ndenumerate() To Display the Data of ndarray for 3-d
import numpy as np
a=np.arange(2*3*3).reshape(2,3,3)
print("3-d Array Data\n{} Dimension={}".format(a,a.ndim))
for val in np.ndenumerate(a):
    print("\t Matrix:{} Row:{} Column:{} Value:{}".format(val[0][0],val[0][1],val[0][2],val[1]))
```

```python
#By Using numpy.ndenumerate() To Display the Data of ndarray for 4-d
import numpy as np
a=np.arange(2*2*3*3).reshape(2,2,3,3)
print("4-d Array Data\n{} Dimension={}".format(a,a.ndim))
for val in np.ndenumerate(a):
    print("\t Cluster:{} Matrix:{} Row:{} Column:{}---->Value:{}".format(val[0][0],val[0][1],val[0][2],val[0][3],val[1]))
```
