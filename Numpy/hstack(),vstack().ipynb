```python
import numpy as np
a=np.array([[10,20,30],[40,50,60],[70,80,90]])
print("Matrix-A")
print(a)
print("Shape of Matrix-A=",a.shape)
b=np.array([[15,25],[35,45],[34,1]])
print("Matrix-B")
print(b)
print("Shape of Matrix-B=",b.shape)
```

```python
#Perform Horizontal Stacking (Consider the Number of Rows)
c=np.hstack((a,b))
print("Matrix-C")
print(c)
print("Shape of Matrix-C=",c.shape)
```

```python
#Perform Horizontal Stacking (Consider the Number of Rows)
d=np.hstack((b,a))
print("Matrix-D")
print(d)
print("Shape of Matrix-D",d.shape)
```

```python
#Creating Two Matrices 
import numpy as np
a=np.array([[10,20,30],[40,50,60],[70,80,90]])
print("Matrix-A")
print(a)
print("Shape of Matrix-A=",a.shape)
b=np.array([[15,25,34],[34,1,55]])
print("Matrix-B")
print(b)
print("Shape of Matrix-B=",b.shape)
```

```python
#Perform Vertical Stacking (Consider the Number of Cols)
c=np.vstack((a,b))
print("Matrix-C")
print(c)
print("Shape of Matrix-C=",c.shape)
```

```python
#Perform Vertical Stacking (Consider the Number of Cols)
d=np.vstack((b,a))
print("Matrix-D")
print(d)
print("Shape of Matrix-D=",d.shape)
```

```python
#Create a Matrix Using diag()
import numpy as np
a=np.arange(9).reshape(3,3)
print("Matrix-A")
print(a)
print("Shape of a={}".format(a.shape))
print("Dimension of a={}".format(a.ndim))
```

```python
#Get The Principle Diagonal Elements
pd=np.diag(a,0)  -----#k=0
print("Principle Diagonal={} Type={}".format(pd,type(pd)))
```

```python
#Get Lower Principle Diagonal Elements
print("Matrix-A")
print(a)
pd=np.diag(a,k=-1)
print("Principle Diagonal={} Type={}".format(pd,type(pd)))
```

```python
#Get Lower Principle Diagonal Elements
print("Matrix-A")
print(a)
pd=np.diag(a,k=1)
print("Principle Diagonal={} Type={}".format(pd,type(pd)))
```
