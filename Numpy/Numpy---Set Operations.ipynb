```python
#Finding Unique Elements from ndarray---numpy.unique()
import numpy as np
a=np.array([1,1,1,2,3,4,5,5,5,6])
b=np.unique(a)
print(b)
```

```python
#Finding the Union of two ndarray Objects--numpy.union1d()
import numpy as np
a=np.array([1,2,3,4])
b=np.array([3,4,5,6])
c=np.union1d(a,b)
print(c)
```

```python
#Finding the Intersection of two ndarray Objects--numpy.intersect1d()
import numpy as np
a=np.array([1,2,3,4])
b=np.array([3,4,5,6])
c=np.intersect1d(a,b)
print(c,type(c))
```

```python
#Finding the Difference  of two ndarray Objects--numpy.setdiff1d()
import numpy as np
a=np.array([1,2,3,4])
b=np.array([3,4,5,6])
c=np.setdiff1d(a,b)
print(c)
```

```python
#Finding the Difference  of two Ndarray Objects--numpy.setdiff1d()
import numpy as np
a=np.array([1,2,3,4])
b=np.array([3,4,5,6])
c=np.setdiff1d(a,b)
print(c,type(c))
```

```python
#Finding the Difference  of two Ndarray Objects--numpy.setdiff1d()
import numpy as np
arr1=np.array([1,2,3,4])
arr2=np.array([3,4,1,2])
newarr=np.setdiff1d(arr1,arr2)
print(newarr,type(newarr))
```

```python
#Finding the symmetric difference ff two Ndarray Objects--numpy.setxor1d()
import numpy as np
arr1=np.array([1,2,3,4])
arr2=np.array([3,4,5,6])
newarr=np.setxor1d(arr1,arr2)
print(newarr)
```
