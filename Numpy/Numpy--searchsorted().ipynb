```python
#Using searchsorted()
import numpy as np
a=np.array([6,7,8,9])
x=np.searchsorted(a,7)
print(x)
```

```python
#Searching from side 
import numpy as np
a=np.array([6,7,8,9])
x=np.searchsorted(a,7,side="right")
print(x)
```

```python
#Unknown Value
import numpy as np
a=np.array([6,7,8,9])  #Value will be taken at last
x=np.searchsorted(a,18)
print(x)
```
