```python
import pandas as pd
# List of Tuples
data1= [('Jack', 34, 'Sydney', 5) ,
        ('Riti', 31, 'Delhi' , 7) ,
        ('Aadi', 46, 'New York', 11)]
# List of Tuples
data2= [('Mohit', 34, 'Tokyo', 11) ,
        ('Veena', 31, 'London' , 10) ,
        ('Shaun', 36, 'Las Vegas', 12)]
# List of Tuples
data3= [('Mark', 47, 'Mumbai',   13) ,
        ('Jose', 43, 'Yokohama', 14) ,
        ('Ramu', 49, 'Paris',    15)]
# Create a DataFrame object from list of tuples
first=pd.DataFrame(data1,
                   columns=["Name","Age","City","Experience"],
                   index=['a','b','c'])
second=pd.DataFrame(data2,
                    columns=["Name","Age","City","Experience"],
                    index=['d','e','f'])
third=pd.DataFrame(data3,
                   columns=["Name","Age","City","Experience"],
                   index=['g','h','i'])
print("First DataFrame")
print(first)
print("---------------------------------------------")
print("Second DataFrame")
print(second)
print("---------------------------------------------")
print("Third DataFrame")
print(third)
```

```python
# Concatenate three DataFrames along the Rows
import pandas as pd
a=pd.concat([first,second,third])
print(a)
```

```python
# Concatenate three DataFrames along the Columns (axis=1)
import pandas as pd
a=pd.concat([first,second,third],axis=1)
print(a)
```

```python
# Concatenate three DataFrames along the Rows (with default axis=0
import pandas as pd
a=pd.concat([first,second,third],axis=0)
print(a)
```
