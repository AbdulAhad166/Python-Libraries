```python
#Create an Object Series  with Default Indices
import numpy as np
import pandas as pd
a=pd.Series(["Rossum","Travis","Kinney","Dennis","Ritche"])
print("Content of Series Object")
print(a,type(a))
```

```python
#Create an Object Series  with  Programmer Defined Indices
import numpy as np
import pandas as pd
a=pd.Series(["Rossum","Travis","Kinney","Dennis","Ritche"],
            index=["Name"+str(i) for i in range(1,len(["Rossum","Travis","Kinney","Dennis","Ritche"])+1)])
print("Content of Series Object")
print(a,type(a))
```

```python
#Find the Dimension of series Object by using 'ndim'
print("Dimension of Series Object a=",a.ndim)
```

```python
#Find the Data type of series Object by using 'dtype'
print("Data Type of Series Object a=",a.dtype)
```

```python
#Find the size and Shape of series Object by using 'size' and 'shape'
print("Number of Elements in Series Object a=",a.size)
print("Shape of Series Object a=",a.shape)
```

```python
#Create an Object Series  with  Programmer Defined Indices with Series Name
import pandas as pd
import numpy as np
a=pd.Series(["Rossum","Travis","Kinney","Dennis","Ritche"],
            index=["Name"+str(i) for i in range(1,len(["Rossum","Travis","Kinney","Dennis","Ritche"])+1)],
           name="Employees")
print("Content of Series Object")
print(a,type(a))
```

```python
#Get the Name of Series Object by using 'name' attribute
print("Name of the Series Object=",a.name)
```

```python
#Get the Indices of Series Object by using 'index' attribute
print("Indices of Series Object=",a.index)
```

```python
for indname in a.index:
    print(indname)
```
