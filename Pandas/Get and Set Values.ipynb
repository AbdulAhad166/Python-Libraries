```python
#Create an Object Series  with  Programmer Defined Indices
import pandas as pd
import numpy as np
a=pd.Series(["Rossum","Travis","Kinney","Dennis","Ritche"],
            index=["Name"+str(i) for i in range(1,len(["Rossum","Travis","Kinney","Dennis","Ritche"])+1)])
print("Content of Series Object")
print(a,type(a))
```

```python
#Create an Object Series  with Default Indices
import pandas as pd
import numpy as np
a=pd.Series(["Rossum","Travis","Kinney","Dennis","Ritche"])
print("Content of Series Object")
print(a,type(a))
```

```python
# Set OR Modify the Value of Series Object By using Indexing
```

```python
print(a,id(a))
```

```python
#Set OR Update 2nd Index Name as "WES MC KINNEY"
a[2]="WES MC KINNEY"
print(a,id(a))
```

```python
# Set OR Modify the Values of Series Object By using Random Indexing
```

```python
#Update 1st and 4th Index names 
a[[1,4]]=["Travis Oliphant","Dennis Ritche"]
print(a,id(a))
```

```python
# Set OR Modify the Values of Series Object By using Slicing
```

```python
print(a,id(a))
```

```python
a[0:3]=["Rossum","Travis","Kinney"]
print(a,id(a))
```
