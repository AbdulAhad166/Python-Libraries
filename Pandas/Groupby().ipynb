```python
import pandas as pd
data = {
    'Department': ['IT', 'IT', 'HR', 'HR', 'IT'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'],
    'Salary': [50000, 60000, 40000, 45000, 70000]
}

a= pd.DataFrame(data)
print(a)
```

```python
for gn,gf in a.groupby(["Department","Gender"]):
    print("Group Name:",gn)
    print(gf)
```

```python
for mg,gf in a.groupby(['Department', 'Gender']):
    print("{}-{}".format(mg[0],mg[1]))
    print(gf)
```

```python
#calculate the average salary for each department and gender combination.
a.groupby(["Department","Gender"])["Salary"].mean()
```

```python
# Using Multiple Aggregations with agg()
result=a.groupby("Department")["Salary"].agg(["sum","mean","max","min","count"])
print(result)
```

```python
import pandas as pd
data = {
    'City': ['Hyderabad', 'Chennai', 'Hyderabad', 'Mumbai', 'Chennai'],
    'Sales': [1000, 2000, 1500, 3000, 2500]
}
a=pd.DataFrame(data)
print(a)
```

```python
for gn,gf in a.groupby("City"):
    print("Group Name:",gn)
    print(gf)
```

```python
#Calculate the total sales for each city.
a.groupby("City")["Sales"].sum()
```

```python
#Calculate the average sales for each city.
a.groupby("City")["Sales"].mean()
```

```python
#Find the maximum sales for each city.
a.groupby("City")["Sales"].max()
```

```python
#Count the number of records for each city.
a.groupby("City").size()
```

```python
#Calculate the total and average sales together using agg().
a.groupby("City")["Sales"].agg(["sum","mean","max","min"])
```

```python
values=a.groupby("City").agg({"Sales":"mean"})
print(values)
```
