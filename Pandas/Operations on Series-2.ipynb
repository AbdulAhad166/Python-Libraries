```python
#Finding Number Occurences of Unique Values from Series Object(Frequency)-----value_counts()
# Create an object of Series 
import pandas as pd
s1=pd.Series([10,20,30,10,20,30,50,60,40,10])
print("Content of s1")
print(s1,type(s1))
```

```python
s1.value_counts()
```

```python
type(s1.value_counts())
```

```python
s1.value_counts().index
```

```python
# Finding Frequences / Occurences
d={}
for val in s1.value_counts().index:
    if val not in d:
        d[val]=1
    else:
        d[val]=d.get(val)+1
print("------------------------------------")
for n,oc in d.items():
    print("\t{}--->{}".format(n,oc))
print("------------------------------------")
```

```python
#Finding Number Occurences of Unique Values from Series Object(Frequency)-----value_counts()
# Create an object of Series 
import pandas as pd
s1=pd.Series([10,20,30,10,20,30,50,60,40,10])
print("Content of s1")
print(s1,type(s1))
```

```python
# Finding Frequences / Occurences
d={}
for val in s1.values:
    if val not in d:
        d[val]=1
    else:
        d[val]=d.get(val)+1
print("------------------------------------")
for n,oc in d.items():
    print("\t{}--->{}".format(n,oc))
print("------------------------------------")
```
