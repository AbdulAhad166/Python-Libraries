```python
#Create an object of DataFrame By using List in List/ tuple in tuple
import pandas as pd
lst=[[10,20,30,40],["RS","BB","AA","DR"]]
a=pd.DataFrame(lst)
print("Content of DataFrame")
print(a,type(a))
```

```python
#Create an object of DataFrame By using List in List/ tuple in tuple
import pandas as pd
lst=[[10,20,30,40],["RS","BB","AA","DR"]]
a=pd.DataFrame(lst,index=["Id1","Id2"],columns=["Rec"+str(i) for i in range(1,len(lst[1])+1)])
print("Content of DataFrame")
print(a,type(a))
```

```python
#Create an object of DataFrame By using List in List/ tuple in tuple
import pandas as pd
lst=[[10,"RS"],(20,"BB"),(30,"AA"),[40,"DR"]]
a=pd.DataFrame(lst,index=["Rec"+str(i) for i in range(1,len(lst)+1)],columns=["ID1","ID2"])
print("Content of DataFrame")
print(a,type(a))
```

```python
#Create DataFrame by using ndarray Object
import numpy as np
import pandas as pd
n=np.array([[10,20,30],[40,50,60],[70,80,90]])
a=pd.DataFrame(n)
print("Content of DataFrame")
print(a,type(a))
```

```python
#Create DataFrame by using ndarray Object
import numpy as np
import pandas as pd
n=np.array([[10,20,30],[40,50,60],[70,80,90]]).flatten()
a=pd.DataFrame(n)
print("Content of DataFrame")
print(a,type(a))
```

```python
#Create DataFrame by using ndarray Object
import numpy as np
import pandas as pd
n=np.array([[10,20,30],[40,50,60],[70,80,90]]).flatten()
s=pd.Series(n)
a=pd.DataFrame(s)
print("--------------------------------")
print("Content of DataFrame")
print("--------------------------------")
print(a,type(a))
print("--------------------------------")
```
