```python
import numpy as np
import pandas as pd
d = {'First Score':[100, 90, np.nan, 95],
        'Second Score': [30, np.nan, 45, 56],
        'Third Score':[52, 40, 80, 98],
        'Fourth Score':[np.nan, np.nan, np.nan, np.nan]}
a=pd.DataFrame(d)
print(a)
```
