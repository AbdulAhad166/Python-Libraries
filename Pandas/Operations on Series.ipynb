```python
#Create an Object of Series with List of Names
import pandas as pd
import numpy as np
a=pd.Series(["Rossum","Travis","Kinney","Hunter","Gosling","Brave","Ritche"],
           index=["Name"+str(i) \
                  for i in range(1,len(["Rossum","Travis",\
                                        "Kinney","Hunter","Gosling","Brave","Ritche"])+1)])
print(a,type(a))
```

```python
#Drop OR Remove the Name which is Index at Name6
a.drop(["Name6"])
```

```python
print(a,type(a))
```

```python
#Drop OR Remove the Name which is Index at Name6
a.drop(["Name6"],inplace=True)
```

```python
print(a,type(a))
```

```python
#Drop OR Remove Multiple Names from Series Object--name1, Name3 and Name7
a.drop(["Name1","Name3","Name7"])
```

```python
#Create an Object of Series with Missing Values
import pandas as pd
s1=pd.Series([None,10,20,30,np.nan,40,pd.NA,50,60,np.nan])
print("Content of s1")
print(s1,type(s1))
```

```python
s1.size
```

```python
#Find the Missing Values from series object by using isna()
print(s1)
print("--------------------------------------")
print(s1.isna(),type(s1.isna()))
```

```python
#Pass Series Boolean Array to Series Object
print("List of Missing Values in Series Object")
s1[s1.isna()]
```

```python
#Create an Object of Series with Missing Values
import pandas as pd
s1=pd.Series([None,10,20,30,np.nan,40,pd.NA,50,60,np.nan])
print("Content of s1")
print(s1,type(s1))
```

```python
#Find the Missing Values from series object by using isnull()
print(s1)
print("----------------------------------")
print(s1.isnull(),type(s1.isnull()))
```

```python
#Pass Series Boolean Array to Series Object
print("List of Missing Values in Series Object")
s1[s1.isnull()]
```
