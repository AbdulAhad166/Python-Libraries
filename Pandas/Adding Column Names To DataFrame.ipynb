```python
import pandas as pd
a=pd.read_csv("F:\\Python-Libraries\\Pandas\\Studentmarks.csv")
print(a)
```

```python
#Compute total marks of all students by adding new col 'total'
a["total"]=0.0
print(a)
```

```python
a["total"]=a["telugu"]+a["english"]+a["hindi"]+a["maths"]+a["science"]+a["social"]
print(a)
```

```python
#Calculate percent by adding new Column 'percent'
a["percent"]=(a["total"]/600)*100
print(a)
```

```python
import numpy as np
a["percent"]=np.around((a["total"]/600)*100,2)
print(a)
```

```python
a["grade"]=None
print(a)
```

```python
#Decide the Grading based the following Conditions
# if percent ranges between 75.0 to 100.0 then Given Distinction
# if percent ranges between 70.0 to 75.0(Excluded) then Given First
# if percent ranges between 60.0 to 70.0(Excluded) then Given Second
# if percent  <60(Excluded) then Given Third
```

```python
# if percent ranges between 75.0 to 100.0 then Given Distinction
a.loc[(a["percent"]<=100)&(a["percent"]>=75),["grade"]]="DISTINCTION"
print(a)
```

```python
# if percent ranges between 70.0 to 75.0(Excluded) then Given First
a.loc[(a["percent"]>=70)&(a["percent"]<75),["grade"]]="FIRST"
print(a)
```

```python
# if percent ranges between 60.0 to 70.0(Excluded) then Given Second
a.loc[(a["percent"]>=60)&(a["percent"]<70),["grade"]]="SECOND"
print(a)
```

```python
# if percent <60(Excluded) then Given Third
a.loc[(a["percent"]<60),["grade"]]="THIRD"
print(a)
```
