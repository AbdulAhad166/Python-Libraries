```python
#Create an Object Series with Numerical Values
import numpy as np
import pandas as pd
a=pd.Series([10,20,30,40,50])
print("Content of Series Object")
print(a,type(a))
```

```python
# Finding the Sum of Elements of Series Object----sum()
print(a,type(a))
print("Sum of Series Values=",a.sum())
```

```python
# Finding Sum of Series of Values
i=0
for val in a.values:
    i=i+val
print("Sum of Series Values=",i)
```

```python
# Finding Number of Non-Missing Values-----------count()
#Create an Object Series with Numerical Values
import numpy as np
import pandas as pd
a=pd.Series([10,None,20,30,pd.NA,40,50,np.nan])
print("Content of Series Object")
print(a,type(a))
```

```python
print("Number of Non-Missing Values=",a.count())
```

```python
print("Number of Missing Values=",a.isna().sum())
```

```python
print("Number of Missing Values=",a.size-a.count())
```

```python
#Finding the median of Elements of Series Object----median()
import numpy as np
import pandas as pd
a=pd.Series([10,2,13,14,4,15,6,16],dtype=float)
print("Content of a")
print("---------------------------------")
print(a,type(a))
print("---------------------------------")
print("Median\n",a.median())
```

```python
#Finding the median of Elements of Series Object----median()
import numpy as np
import pandas as pd
a=pd.Series([10,2,13,14,4,15,16],dtype=float)
print("Content of a")
print("---------------------------------")
print(a,type(a))
print("---------------------------------")
print("Median\n",a.median())
```

```python
#Finding the variance and standard Deviation of Elements of Series Object----var(),std()
import numpy as np
import pandas as pd
a=pd.Series([10,2,13,14,4,15,16],dtype=float)
print("Content of a")
print("---------------------------------")
print(a,type(a))
print("---------------------------------")
print("Variance=",a.var())
print("Standard Deviation=",a.std())
```

```python
# Finding the Aggregation of Statistical Operations of Series Object---agg()----MOST IMP
import numpy as np
import pandas as pd
a=pd.Series([10,2,13,14,4,15,16],dtype=float)
print("Content of a")
print("---------------------------------")
print(a,type(a))
print("---------------------------------")
print(a.agg(["mean","median","var","std"]))
```

```python
# Finding the Aggregation of Statistical Operations of Series Object---agg()
import numpy as np
import pandas as pd
a=pd.Series([10,2,13,14,4,15,16],dtype=float)
print("Content of a")
print("---------------------------------")
print(a,type(a))
print("---------------------------------")
print(a.agg(["max","min","sum","count"]))
```

```python
# Finding the Statistical Information of   of Elements of Series Object	(memory layout)-----info()
import numpy as np
import pandas as pd
a=pd.Series([None,10,2,13,pd.NA,14,4,15,16],name="NIT")
print("Content of a")
print("---------------------------------")
print(a,type(a))
print("---------------------------------")
print("Statistical Information of Elements of Series Object--with Memory Usage")
print("------------------------------------------------------------------------")
print(a.info())
```

```python
#Finding the Product of Elements of Series Object---------------product()
import numpy as np
import pandas as pd
a=pd.Series([1,2,3,4,5])
print("Content of a")
print("---------------------------------")
print(a,type(a))
print("---------------------------------")
print("Product=",a.product())
```
