```python
#Type Casting Perticular Column Values of DataFrame--astype()
import numpy as np
import pandas as pd
tech={"Fees":['20000','25000','2600'],"Discount":['1000','2300','1500']}
a=pd.DataFrame(tech)
print(a,type(a))
```

```python
a.dtypes
```

```python
#Type Cast the Values of 'Fee' Column Data
a["Fees"]=a["Fees"].astype(int)
```

```python
a.dtypes
```

```python
#Type Cast the Values of 'Discount' Column Data
a["Discount"]=a["Discount"].astype("f2")
```

```python
a.dtypes
```

```python
#Type Casting Perticular Column Values of DataFrame--astype()
import numpy as np
import pandas as pd
tech={"Courses":["Python","Django","DSA"],"Fees":['20000','25000','26000'],"Discount":['1000','2300','1500']}
a=pd.DataFrame(tech)
print(a,type(a))
```

```python
a.dtypes
```

```python
#Type Cast Multiple Colmun Values at a time with Different types
a=a.astype({"Courses":"str","Fees":"int","Discount":"float"})
```

```python
a.dtypes
```

```python
a=a.astype({"Fees":"i2","Discount":"f2"})
```
