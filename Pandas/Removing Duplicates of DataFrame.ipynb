```python
import pandas as pd
a=pd.read_csv("F:\\Python-Libraries\\Pandas\\Studentmarks.csv")
print(a)
```

```python
a["total"]=a["telugu"]+a["english"]+a["hindi"]+a["maths"]+a["science"]+a["social"]
print(a)
```

```python
import numpy as np
a["percent"]=np.around((a["total"]/600)*100,2)
print(a)
```

```python
#Sort the DataFrame Data Based on Col Name---percent
a.sort_values(["percent"])
```
