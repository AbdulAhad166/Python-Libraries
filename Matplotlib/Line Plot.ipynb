```python
#Draw the Line Plot for the following
#       No. of Days:	1    2    3    4     5     6
#		Absenties:		4   7     2    10    3     8
```

```python
import matplotlib.pyplot as plt
days=[1,2,3,4,5,6]
Absenties=[4,7,2,10,3,8]
plt.plot(days,Absenties)
plt.show()
```

```python
import matplotlib.pyplot as plt
days=[1,2,3,4,5,6]
absenties=[4,7,2,10,3,8]
plt.title("Number of Days Vs Absenties")
plt.xlabel("Number of Days")
plt.ylabel("Absenties")
plt.plot(days,absenties)
plt.show()
```

```python
#Add Some Effects to the Line Plot
# Line Properties, Marker Properties
```

```python
#Line Properties
import matplotlib.pyplot as plt
days=[1,2,3,4,5,6]
absenties=[4,7,2,10,3,8]
plt.plot(days,absenties,ls="dotted",lw=5,c="black")
plt.title("Number of Days Vs Absenties")
plt.xlabel("Number of Days")
plt.ylabel("Absenties")
plt.show()
```

```python
#Line Properties
import matplotlib.pyplot as plt
days=[1,2,3,4,5,6]
absenties=[4,7,2,10,3,8]
plt.plot(days,absenties,ls="-.",lw=3,c="green")
plt.title("Number of Days Vs Absenties")
plt.xlabel("Number of Days")
plt.ylabel("Absenties")
plt.show()
```

```python
#Marker Properties
import matplotlib.pyplot as plt
days=[1,2,3,4,5,6]
absenties=[4,7,2,10,3,8]
plt.plot(days,absenties,marker=".",mec="black",mfc="yellow",ms=10)
plt.title("Number of Days Vs Absenties")
plt.xlabel("Number of Days")
plt.ylabel("Absenties")
plt.show()
```

```python
#Marker and Line Properties
import matplotlib.pyplot as plt
days=[1,2,3,4,5,6]
absenties=[4,7,2,10,3,8]
plt.plot(days,absenties,lw=3,c="black",marker="d",mec="red",mfc="yellow",ms=10)
plt.title("Number of Days Vs Absenties")
plt.xlabel("Number of Days")
plt.ylabel("Absenties")
plt.show()
```

```python
#Grid Properties,Marker Properties,Line Properties 
import matplotlib.pyplot as plt
days=[1,2,3,4,5,6]
absenties=[4,7,2,10,3,8]
plt.plot(days,absenties,lw=3,c="green",marker=".",mec="black",mfc="yellow",ms=10)
plt.title("Number of Days Vs Absenties")
plt.xlabel("Number of Days")
plt.ylabel("Absenties")
plt.grid()
plt.show()
```

```python
#Grid Properties,Marker Properties,Line Properties 
import matplotlib.pyplot as plt
days=[1,2,3,4,5,6]
absenties=[4,7,2,10,3,8]
plt.plot(days,absenties,lw=3,c="green",marker=".",mec="black",mfc="yellow",ms=10)
plt.title("Number of Days Vs Absenties")
plt.xlabel("Number of Days")
plt.ylabel("Absenties")
plt.grid(axis="both",ls="dotted",lw=1,c="violet")
plt.show()
```

```python
#Adding the grid Properties along with Marker and line  Properties 
#Formating x-label, y-label and Title with fontfamily values in the form of fontdict
days=[1,2,3,4,5,6]
absenties=[4,7,2,10,3,8]
plt.plot(days,absenties,lw=3,c="green",marker=".",mec="black",mfc="yellow",ms=10)
plt.title("Number of Days Vs Absenties",fontdict={"family":"arial","size":"15","color":"brown"})
plt.xlabel("Number of Days",fontdict={"family":"arial","size":"10","color":"red"})
plt.ylabel("Absenties",fontdict={"family":"arial","size":"10","color":"red"})
plt.grid(axis="both",ls="dotted",lw=1,c="violet")
plt.show()
```

```python
#Adding the grid Properties along with Marker and line  Properties 
#Formating x-label, y-label and Title with fontfamily values in the form of fontdict
import matplotlib.pyplot as plt
import numpy as np
days=np.array([1,2,3,4,5,6])
absenties=np.array([4,7,2,10,3,8])
plt.plot(days,absenties,lw=3,c="green",marker=".",mec="black",mfc="yellow",ms=10)
plt.title("Number of Days Vs Absenties",fontdict={"family":"arial","size":"15","color":"brown"})
plt.xlabel("Number of Days",fontdict={"family":"arial","size":"10","color":"red"})
plt.ylabel("Absenties",fontdict={"family":"arial","size":"10","color":"red"})
plt.grid(axis="both",ls="dotted",lw=1,c="violet")
plt.show()
```

```python
#Adding the grid Properties along with Marker and line  Properties 
#Formating x-label, y-label and Title with fontfamily values in the form of fontdict
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
days=pd.Series(np.array([1,2,3,4,5,6]))
absenties=pd.Series(np.array([4,7,2,10,3,8]))
plt.plot(days,absenties,lw=3,c="green",marker=".",mec="black",mfc="yellow",ms=10)
plt.title("Number of Days Vs Absenties",fontdict={"family":"arial","size":"15","color":"brown"})
plt.xlabel("Number of Days",fontdict={"family":"arial","size":"10","color":"red"})
plt.ylabel("Absenties",fontdict={"family":"arial","size":"10","color":"red"})
plt.grid(axis="both",ls="dotted",lw=1,c="violet")
plt.show()
```
