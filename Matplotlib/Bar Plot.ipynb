```python
import matplotlib.pyplot as plt
import pandas as pd
empdata=pd.read_csv("F:\\Python-Libraries\\Matplotlib\\emp_save.csv")
plt.bar(empdata["Savings"],empdata["NumberOfEmployees"])
plt.show()
```

```python
from matplotlib import pyplot as plt
x=["Apples","Kiwi","Sberry","Mango"]
y=[40,20,35,23]
colors=["red","brown","hotpink","yellow"]
labels=["Apples","Kiwi","Sberry","Mango"]
plt.bar(x,y,color=colors,width=0.5,label=labels)
plt.legend()
plt.xlabel("Fruits Types")
plt.ylabel("Number of Units Sold-Out")
plt.title("Fruits Selling Analysis")
plt.show()
```
