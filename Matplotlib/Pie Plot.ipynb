```python
from matplotlib import pyplot as plt
plt.title("Games",color="green",fontweight="bold",fontsize=20)
numbers=[10,5,5,10,10]
sports=["Football","Hockey","Cricket","Basketball","Batminton"]
color=["red","blue","green","grey","brown"]
plt.pie(numbers,labels=sports,explode=[0.1,0,0,0,0],colors=color,startangle=90,autopct='%0.0f%%',wedgeprops={'linewidth':4,'edgecolor':"black"})
plt.show()
```

```python
from matplotlib import pyplot as plt
plt.title("Games View",color="green",fontweight="bold",fontsize=20)
numbers=[10,5,5,10,10]
sports=["FootBall","Hockey","Cricket","Basketball","Badminton"]
color=["red","blue","green","grey","brown"]
plt.pie(numbers,labels=sports,explode=[0.1,0,0,0,0],colors=color,startangle=0,autopct='%0.1f%%',
        wedgeprops={'linewidth':4,'edgecolor':"black"})
plt.show()
```

```python
#Question: Nobita lists down her monthly expenditure as follows:
#			Expenditure		Amount
#			Rent			4000
#			Food			5400
#			Clothing			2800
#			Savings			400
#		Draw a pie chart for her monthly expenses.
```

```python
from matplotlib import pyplot as plt
plt.title("Nobita Expenditure",color="black",fontweight="bold",fontsize=10)
expense=[4000,5400,2800,400]
expname=["Rent","Food","Clothing","Savings"]
color=["red","blue","green","grey","brown"]
plt.pie(expense,labels=expname,explode=[0,0,0,0.1],colors=color,autopct='%0.0f%%',
        wedgeprops={'linewidth':4,'edgecolor':"black"})
plt.show()
```
